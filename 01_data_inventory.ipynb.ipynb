```python
officers      = pd.read_csv("/Users/nanikokakonashvili/Downloads/offshore_leaks_project/data/raw/full-oldb.LATEST/nodes-officers.csv",     low_memory=False)
entities      = pd.read_csv("/Users/nanikokakonashvili/Downloads/offshore_leaks_project/data/raw/full-oldb.LATEST/nodes-entities.csv",     low_memory=False)
relationships = pd.read_csv("/Users/nanikokakonashvili/Downloads/offshore_leaks_project/data/raw/full-oldb.LATEST/relationships.csv",      low_memory=False)
 
## Quick sanity check
print("Officers:",      len(officers))
print("Entities:",      len(entities))
print("Relationships:", len(relationships))
```

```python
def get_entities_for_country(code):
   
    ##Returns a DataFrame of offshore entities linked to
   ## officers whose country_codes contains `code`.
   ###
    ## Step 1: filter officers by country
    country_officers = officers[
        officers["country_codes"].str.contains(code, na=False)
    ]
 
    ## Step 2: inner-join to relationships on node_id
    ##         keep only officer_of links
    linked = relationships[
        (relationships["node_id_start"].isin(country_officers["node_id"])) &
        (relationships["rel_type"] == "officer_of")
    ]
 
    ## Step 3: inner-join to entities
    return entities[entities["node_id"].isin(linked["node_id_end"])]
 
```

```python
country_map = {
    "GEO": "Georgia",
    "RUS": "Russia",
    "DEU": "Germany",
    "FRA": "France"
}
 
## Jurisdictions we care about
top_juris = [
    "British Virgin Islands", "Malta", "Seychelles",
    "Bahamas", "Bermuda", "Cayman Islands", "Samoa", "Panama"
]
 
## Build a dict of {country_label: {jurisdiction: pct}}
pct_data = {}
totals   = {}
 
for code, label in country_map.items():
    ents          = get_entities_for_country(code)
    total         = len(ents)
    totals[label] = total
    ## Use a dict comprehension — covered in user-defined functions week
    pct_data[label] = {
        j: round((ents["jurisdiction_description"] == j).sum() / total * 100, 1)
        if total > 0 else 0
        for j in top_juris
    }
 
## Convert to DataFrame for easy inspection
pct_df = pd.DataFrame(pct_data).T          ## rows = countries, cols = jurisdictions
print("\nJurisdiction % by country:")
print(pct_df)
```

```python
## Technique: plt.bar with manual x-offset per group (class slide pattern)
 
## Abbreviated x-axis labels
short_labels = ["BVI", "Malta", "Seychelles", "Bahamas",
                "Bermuda", "Cayman", "Samoa", "Panama"]
 
## Colors per country
colors = ["#c9a84c", "#c0392b", "#3498db", "#2ecc71"]
 
n_juris   = len(top_juris)
n_ctries  = len(country_map)
x         = np.arange(n_juris)        ## base x positions
width     = 0.18                       ## width of each bar
 
plt.figure(figsize=(13, 7))
 
## Loop over countries — same pattern as class bar chart examples
for i, (code, label) in enumerate(country_map.items()):
    ## Offset each country's bars so they sit side by side
    offset = (i - (n_ctries - 1) / 2) * width
    vals   = [pct_data[label][j] for j in top_juris]
 
    plt.bar(
        x + offset,
        vals,
        width,
        label=f"{label}  (n={totals[label]:,})",
        color=colors[i],
        alpha=0.88,
        edgecolor="white",
        linewidth=0.5
    )
 
    ## Add value labels on top of bars > 2% (plt.text, covered in class)
    for xi, val in zip(x + offset, vals):
        if val >= 2.5:
            plt.text(
                xi, val + 0.3,
                f"{val:.0f}%",
                ha="center", va="bottom",
                fontsize=7, color=colors[i], fontweight="bold"
            )
 
## Labels, ticks, legend — exactly as shown in class slides
plt.xticks(x, short_labels, fontsize=11)
plt.yticks(fontsize=10)
plt.xlabel("Offshore Jurisdiction", fontsize=12)
plt.ylabel("Share of linked entities (%)", fontsize=12)
plt.title(
    "Offshore Jurisdiction Preferences by Officer Nationality\n"
    "ICIJ Offshore Leaks Database — all five leaks combined",
    fontsize=13, fontweight="bold", loc="left"
)
plt.legend(fontsize=10)
 
## Key annotation highlighting Georgia's hybrid pattern
plt.annotate(
    "Georgia: high BVI (46%) like Russia,\n"
    "but also high Malta (26%) — closer to EU.\n"
    "Suggests a hybrid offshore strategy.",
    xy=(1, pct_data["Georgia"]["Malta"]),      ## arrow points at Georgia's Malta bar
    xytext=(3.0, 35),
    fontsize=9,
    color="#c9a84c",
    arrowprops=dict(arrowstyle="->", color="#c9a84c", lw=1.2),
    bbox=dict(boxstyle="round,pad=0.4", facecolor="lightyellow",
              edgecolor="#c9a84c", alpha=0.9)
)
 
plt.tight_layout()
plt.savefig("viz1_jurisdiction_profile.png", bbox_inches="tight", dpi=150)
plt.show()
print("Viz 1 saved.")
```

```python
## Get Georgian-linked entities
geo_ents = get_entities_for_country("GEO").copy()
 
## Parse incorporation year — pd.to_datetime covered in Pandas sessions
geo_ents["inc_year"] = pd.to_datetime(
    geo_ents["incorporation_date"], format="mixed", errors="coerce"
).dt.year
 
## Classify jurisdictions by opacity level
opaque_havens = {
    "British Virgin Islands", "Seychelles", "Bahamas",
    "Panama", "Samoa", "Cook Islands", "Belize"
}
semi_transparent = {
    "Malta", "Bermuda", "Isle of Man", "Cayman Islands", "Mauritius"
}
 
## User-defined function (class skill) to classify each jurisdiction
def classify_jurisdiction(juris):
    if juris in opaque_havens:
        return "Opaque haven"
    elif juris in semi_transparent:
        return "Semi-transparent"
    else:
        return "Undetermined"
 
## Apply the function — .apply() pattern from class
geo_ents["juris_type"] = geo_ents["jurisdiction_description"].apply(
    classify_jurisdiction
)
 
## Groupby + unstack to get year × type counts — groupby covered in class
timeline = (
    geo_ents[geo_ents["inc_year"].between(1991, 2018)]
    .groupby(["inc_year", "juris_type"])
    .size()
    .unstack(fill_value=0)
    .reindex(range(1991, 2019), fill_value=0)
)
 
## Make sure all columns exist even if empty
for col in ["Opaque haven", "Semi-transparent", "Undetermined"]:
    if col not in timeline.columns:
        timeline[col] = 0
 
print("\nTimeline data (first 5 rows):")
print(timeline.head())
```

```python
## Technique: stacked plt.bar + plt.vlines + plt.annotate (class slides)
 
years = timeline.index.astype(int)
 
plt.figure(figsize=(13, 6))
 
## Bottom layer: opaque havens
plt.bar(years, timeline["Opaque haven"],
        color="#c0392b", alpha=0.85,
        label="Opaque haven  (BVI, Seychelles, Panama…)")
 
## Middle layer: semi-transparent — use bottom= to stack
plt.bar(years, timeline["Semi-transparent"],
        bottom=timeline["Opaque haven"],
        color="#3498db", alpha=0.80,
        label="Semi-transparent  (Malta, Bermuda…)")
 
## Top layer: undetermined
plt.bar(years, timeline["Undetermined"],
        bottom=timeline["Opaque haven"] + timeline["Semi-transparent"],
        color="#888888", alpha=0.60,
        label="Undetermined / Other")
 
## Key event lines — plt.vlines + plt.annotate (shown in class slides)
events = {
    2004: "Ivanishvili\nreturns\nto Georgia",
    2012: "Georgian Dream\nwins election",
    2016: "Panama\nPapers\nreleased",
}
 
for yr, label in events.items():
    plt.vlines(yr, 0, 20, colors="#c9a84c",
               linewidth=1.2, linestyle="--", alpha=0.8)
    plt.annotate(
        label,
        xy=(yr, 17),
        xytext=(yr + 0.15, 17),
        fontsize=7.5, color="#c9a84c", style="italic",
        bbox=dict(facecolor="white", edgecolor="none", alpha=0.5, pad=1)
    )
 
## Peak annotation — find the year with most opaque entities
peak_yr  = int(timeline["Opaque haven"].idxmax())
peak_val = int(timeline.loc[peak_yr, "Opaque haven"])
 
plt.annotate(
    f"Peak: {peak_val} new\nopaque entities\nin {peak_yr}",
    xy=(peak_yr, peak_val),
    xytext=(peak_yr + 1.5, peak_val + 1.5),
    fontsize=9, color="#c0392b",
    arrowprops=dict(arrowstyle="->", color="#c0392b", lw=1.2),
    bbox=dict(boxstyle="round,pad=0.4", facecolor="#fff0f0",
              edgecolor="#c0392b", alpha=0.9)
)
 
## Labels, ticks, legend — same class pattern
plt.xlabel("Incorporation Year", fontsize=12)
plt.ylabel("New offshore entities", fontsize=12)
plt.xticks(years[::2], fontsize=9, rotation=45)   ## every other year to avoid crowding
plt.yticks(fontsize=10)
plt.xlim(1990, 2019)
plt.title(
    "Georgian-linked Offshore Entity Incorporations by Year\n"
    "ICIJ Offshore Leaks Database — coloured by jurisdiction type",
    fontsize=13, fontweight="bold", loc="left"
)
plt.legend(fontsize=10, loc="upper left")
 
plt.tight_layout()
plt.savefig("viz2_georgian_timeline.png", bbox_inches="tight", dpi=150)
plt.show()
print("Viz 2 saved.")
 
```
